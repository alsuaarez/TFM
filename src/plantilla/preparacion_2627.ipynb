```python
def evaluar_estatus(row):
    if row['Jugador'] in jugadores_con_contrato:
        return 'Se queda'
    elif row['Jugador'] in ['Lorenzo Amatucci', 'Estanis Pedrola', 'Taisei Miyashiro', 'Iker Bravo']:
        return 'Fin de cesión'
    elif row['Jugador'] in ['José Antonio Caro', 'Mika Mármol', 'Sergio Barcia', 'Enzo Loiodice', 'Nicolás Benedetti', 'Jonathan Viera', 'Jesé', 'Pejiño', 'Sandro Ramírez']:
        # Estos jugadores aparecen en la 25/26 pero no están garantizados con contrato en la 26/27 según los documentos
        return 'Fin de contrato'
    else:
        return 'Revisión'
```

## Jugadores con opción de compra que parece que se ejerce

```python
# Función para ejercer compra de un jugador
def ejercer_opcion_compra(nombre_jugador, df_origen, df_destino):
    # Localizar al jugador en la base de datos origen
    jugador_data = df_origen[df_origen['Jugador'] == nombre_jugador]
    
    if not jugador_data.empty:
        # Añadir al df destino usando pd.concat
        df_destino = pd.concat([df_destino, jugador_data], ignore_index=True)
        
        # Actualizar la variable de estatus para que quede documentado el movimiento
        df_destino.loc[df_destino['Jugador'] == nombre_jugador, 'Estatus_26_27'] = 'Compra ejercida'
        
        print(f"Éxito: {nombre_jugador} ha sido transferido a df_26_27.")
    else:
        print(f"Error: El jugador '{nombre_jugador}' no se encuentra en df_plantilla.")
        
    return df_destino
```

```python
df_26_27 =ejercer_opcion_compra('José Antonio Caro', df_plantilla, df_26_27)
df_26_27 =ejercer_opcion_compra('Taisei Miyashiro', df_plantilla, df_26_27)
df_26_27 =ejercer_opcion_compra('Sandro Ramírez', df_plantilla, df_26_27)
df_26_27 =ejercer_opcion_compra('Jesé', df_plantilla, df_26_27)
df_26_27 =ejercer_opcion_compra('Enzo Loiodice', df_plantilla, df_26_27)
```

## Operación salidas

```python
# Función para vender un jugador
def vender_jugador(nombre_jugador, df_plantilla):
    # Comprobar si el jugador realmente está en el DataFrame actual
    if nombre_jugador in df_plantilla['Jugador'].values:
        
        # Extraer la posición para tener un registro claro de qué hueco queda libre
        posicion = df_plantilla.loc[df_plantilla['Jugador'] == nombre_jugador, 'Posición'].values[0]
        
        # Filtrar el DataFrame: nos quedamos con todos los que NO sean el jugador vendido
        df_plantilla = df_plantilla[df_plantilla['Jugador'] != nombre_jugador].reset_index(drop=True)
        
        # Confirmación visual en consola
        print(f"Venta del jugador: {nombre_jugador} ({posicion})")
            
    else:
        print(f"Error: No se puede vender a '{nombre_jugador}' porque no figura en la plantilla actual.")
        
    return df_plantilla
```

## Guardar plantilla para la carpeta data

```python
# Función para guardar la plantilla en data
def guardar_plantilla_2627(df):
    # Obtener la ruta del directorio de trabajo actual
    ruta_actual = Path.cwd()
    
    # Navegar por el directorio
    ruta_destino = ruta_actual.parent.parent / 'data' / 'udlaspalmas'
    
    # Crear los directorios si aún no existen
    ruta_destino.mkdir(parents=True, exist_ok=True)
    
    # Definir la ruta completa del archivo
    ruta_archivo = ruta_destino / 'df_plantilla_2627.csv'
    
    # Exportar el DataFrame a formato CSV
    df.to_csv(ruta_archivo, index=False, encoding='utf-8-sig')
    
    print(f"El archivo ha sido guardado en: {ruta_archivo}")
```

```python
# Plantilla que se preveé a fecha 18/06/2026
guardar_plantilla_2627(df_26_27)
```
