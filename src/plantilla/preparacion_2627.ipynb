## Creación del DF plantilla de la 25/26 a la 26/27

```python
# Datos extrídos de TranferMarkt de la plantilla 1er equipo
datos_25_26 = {
    'Jugador': [
        'Dinko Horkas', 'José Antonio Caro', 'Adri Suárez', 'Mika Mármol', 
        'Enrique Clemente', 'Sergio Barcia', 'Juanma Herzog', 'Álex Suárez', 
        'Cristian Gutiérrez', 'Viti Rozada', 'Marvin Park', 'Lorenzo Amatucci', 
        'Enzo Loiodice', 'Iñaki González', 'Nicolás Benedetti', 'Kirian Rodríguez', 
        'Iván Gil', 'Jonathan Viera', 'Jeremía Recoba', 'Estanis Pedrola', 
        'Jesé', 'Manu Fuster', 'Taisei Miyashiro', 'Ale García', 'Pejiño', 
        'Iker Bravo', 'Sandro Ramírez', 'Edward Cedeño', 'Aboubacar Bassinga', 
        'Adam Arvelo', 'Iván Cédric'
    ],
    'Posición': [
        'Portero', 'Portero', 'Portero', 'Defensa central', 
        'Defensa central', 'Defensa central', 'Defensa central', 'Defensa central', 
        'Lateral izquierdo', 'Lateral derecho', 'Lateral derecho', 'Pivote', 
        'Mediocentro', 'Mediocentro', 'Mediocentro ofensivo', 'Mediocentro ofensivo', 
        'Mediocentro ofensivo', 'Mediocentro ofensivo', 'Extremo izquierdo', 'Extremo izquierdo', 
        'Extremo izquierdo', 'Extremo derecho', 'Extremo derecho', 'Extremo derecho', 'Extremo derecho',
        'Delantero centro', 'Delantero centro', 'Pivote', 'Mediocentro', 
        'Extremo derecho', 'Delantero centro'
    ],
    'Edad': [27, 32, 21, 24, 27, 25, 22, 33, 25, 28, 25, 22, 25, 21, 29, 30, 26, 36, 22, 22, 33, 28, 26, 23, 29, 21, 30, 22, 20, 21, 24],
    'Valor_Mercado_MEur': [2.5, 0.4, 0.2, 10.0, 1.5, 1.5, 1.5, 0.6, 0.8, 1.5, 1.2, 5.0, 1.2, 0.2, 2.0, 1.0, 0.8, 0.3, 1.2, 0.9, 0.3, 2.0, 1.8, 1.5, 0.6, 4.0, 0.7, 0.3, 0.2, 0.05, 1.0]
}

df_plantilla = pd.DataFrame(datos_25_26)
```

## Estado de los jugadores

```python
def asignar_estatus_contractual(nombre_jugador):
    """Clasifica a un jugador según su vínculo para la temporada 26/27."""
    if nombre_jugador in JUGADORES_CON_CONTRATO:
        return 'Se queda'
    elif nombre_jugador in CESIONES_FINALIZADAS:
        return 'Fin de cesión'
    elif nombre_jugador in CONTRATOS_VENCIDOS:
        return 'Fin de contrato'
    return 'Sin clasificar'


df_plantilla['Estatus_26_27'] = df_plantilla['Jugador'].apply(asignar_estatus_contractual)
df_plantilla
```

## Plantilla base: jugadores con contrato

```python
# ¿Cuántos jugadores por posición se quedan? + Número total de jugadores
df_base_2627 = df_plantilla[df_plantilla['Estatus_26_27'] == 'Se queda'].reset_index(drop=True)

print(f"Jugadores con contrato para la 26/27: {len(df_base_2627)}")
print("\nDistribución por posición:")
print(df_base_2627.groupby('Posición')['Jugador'].count().to_string())
```

## Jugadores con opción de compra que se ejerce y renovaciones

```python
def ejercer_opcion_compra(nombre_jugador, df_origen, df_destino):
    """
    Mueve a un jugador al DataFrame destino y lo marca como 'Compra ejercida'.
    Formaliza el paso de cesión a fichaje definitivo en el registro de plantilla.
    """
    jugador_fila = df_origen[df_origen['Jugador'] == nombre_jugador]

    if jugador_fila.empty:
        print(f"  AVISO: '{nombre_jugador}' no está en la base de datos origen.")
        return df_destino

    df_destino = pd.concat([df_destino, jugador_fila], ignore_index=True)
    df_destino.loc[df_destino['Jugador'] == nombre_jugador, 'Estatus_26_27'] = 'Compra ejercida'

    print(f"  OK: {nombre_jugador} ({jugador_fila['Posición'].values[0]})")
    return df_destino


print("Opciones ejercidas y renovaciones confirmadas:")
for jugador in OPCIONES_EJERCIDAS:
    df_base_2627 = ejercer_opcion_compra(jugador, df_plantilla, df_base_2627)
```

## Operación salidas

```python
def dar_baja(nombre_jugador, df_plantilla):
    """
    Elimina a un jugador de la plantilla y deja constancia de la operación.
    Aplica tanto para ventas como para cesiones salientes o no renovaciones.
    """
    if nombre_jugador not in df_plantilla['Jugador'].values:
        print(f"  AVISO: '{nombre_jugador}' no figura en la plantilla actual.")
        return df_plantilla

    posicion = df_plantilla.loc[df_plantilla['Jugador'] == nombre_jugador, 'Posición'].values[0]
    df_actualizado = df_plantilla[df_plantilla['Jugador'] != nombre_jugador].reset_index(drop=True)

    print(f"  Baja: {nombre_jugador} ({posicion})")
    return df_actualizado
```

```python
# Llamada de la función para una salida
df_base_2627 = dar_baja('Iván Cédric', df_base_2627)
```

## Plantilla base 26/27 - resumen

```python
# Resumen de la plantilla base para la temporada
print(f"Total de jugadores: {len(df_base_2627)}")
print(f"Valor total de mercado: {df_base_2627['Valor_Mercado_MEur'].sum():.1f}M €")
print(f"Media de edad: {df_base_2627['Edad'].mean():.1f} años\n")

resumen = df_base_2627.groupby('Posición').agg(
    Jugadores=('Jugador', 'count'),
    Valor_Total=('Valor_Mercado_MEur', 'sum')
).round(1)

print("Cobertura por posición:")
print(resumen.to_string())
```

## Exportar datos

```python
def guardar_plantilla_2627(df):
    """Guarda la plantilla en la carpeta de datos del proyecto."""
    ruta_actual = Path.cwd()
    ruta_destino = ruta_actual.parent.parent / 'data' / 'udlaspalmas'
    ruta_destino.mkdir(parents=True, exist_ok=True)

    ruta_archivo = ruta_destino / 'df_plantilla_2627.csv'
    df.to_csv(ruta_archivo, index=False, encoding='utf-8-sig')

    print(f'Plantilla guardada en: {ruta_archivo}')
    print(f'Registros exportados: {len(df)}')


# Plantilla prevista a fecha 18/06/2026
guardar_plantilla_2627(df_base_2627)
```
